# Hälsostudie del 2
Syftet med denna analys är att undersöka samband mellan olika hälsovariabler med hjälp av statistiska metoder, korrelationsanalys och multipel regression.

```python
import pandas as pd

from src.health_analyzer import HealthAnalyzer

df = pd.read_csv("data/health_study_dataset.csv")

analyzer = HealthAnalyzer(df)
```

# Metodval

Korrelationsanalys används för att undersöka styrkan och riktningen på samband mellan de numeriska variablerna i datamaterialet.

Multipel linjär regression väljs eftersom metoden gör det möjligt att undersöka hur en variabel påverkar en annan variabel. I denna studie används ålder för att förklara variationer i systoliskt blodtryck.

Regressionen implementerades med hjälp av scikit-learn och visualiseringarna skapades med matplotlib och seaborn.

# Grundläggande Statistik

```python
analyzer.summary_statistics()
```

# Medelålder och Multipel Regression

```python
print("Medelålder:")
print(analyzer.calculate_mean_age())

print("\nRegression R²:")
print(analyzer.blood_pressure_regression())
```

# Resultat av Regressionsanalysen

En multipel linjär regressionsanalys genomfördes för att undersöka hur ålder och vikt påverkar systoliskt blodtryck.

Den framtagna modellen gav ett R²-värde på 0,369. Detta innebär att cirka 36,9 % av variationen i systoliskt blodtryck kan förklaras av variablerna ålder och vikt.

Resultatet tyder på att dessa faktorer har en tydlig påverkan på blodtrycket, men att även andra faktorer som inte ingår i modellen sannolikt bidrar till variationen.

# Korrelationsanalys

```python
analyzer.correlation_heatmap()
```

# Tolkning av korrelationsanalysen

Korrelationsmatrisen visar flera positiva samband mellan variablerna i datasetet.

Det starkaste sambandet återfinns mellan ålder och systoliskt blodtryck (r = 0,61), vilket indikerar att blodtrycket tenderar att öka med stigande ålder.

Det finns även ett positivt samband mellan ålder och kolesterol (r = 0,49).

Resultaten tyder på att ålder är en viktig faktor som påverkar flera hälsorelaterade variabler i datamaterialet.

Analysen visar att ålder har ett tydligt samband med systoliskt blodtryck.

Regressionsmodellen förklarar cirka 36,9 % av variationen i blodtrycket.

Korrelationsanalysen visar att ålder har det starkaste sambandet med systoliskt blodtryck i datasetet.

Både korrelationsmatrisen och regressionsgrafen stödjer slutsatsen att ålder är en viktig faktor som påverkar blodtrycket.
